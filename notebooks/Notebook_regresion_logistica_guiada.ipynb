# Sesión 2 — Regresión logística guiada

**Objetivo:** construir una regla de clasificación sencilla a mano y luego compararla con un modelo de regresión logística en Python.

Usaremos:
- `x1` = horas de estudio
- `x2` = horas de sueño
- `y` = 1 si aprueba, 0 si no aprueba


```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

df = pd.DataFrame({
    "x1_estudio": [1, 2, 3, 4, 5, 6],
    "x2_sueno": [8, 7, 6, 5, 4, 6],
    "aprueba": [0, 0, 0, 1, 1, 1]
})

df
```

## 1. Regla manual

En la hoja de trabajo probaste una regla como:

\[
z = 2x_1 - x_2 - 1
\]

y después decidiste:
- si `z > 0`, clase 1
- si `z <= 0`, clase 0


```python
df["z_manual"] = 2*df["x1_estudio"] - df["x2_sueno"] - 1
df["pred_manual"] = (df["z_manual"] > 0).astype(int)
df
```

```python
print("Accuracy regla manual =", accuracy_score(df["aprueba"], df["pred_manual"]))
print("Matriz de confusión:")
print(confusion_matrix(df["aprueba"], df["pred_manual"]))
```

## 2. Modelo de regresión logística

Ahora dejamos que el algoritmo ajuste automáticamente los coeficientes.


```python
X = df[["x1_estudio", "x2_sueno"]]
y = df["aprueba"]

model = LogisticRegression()
model.fit(X, y)

print("Coeficientes:", model.coef_)
print("Intercepto:", model.intercept_)
```

```python
df["prob_aprueba"] = model.predict_proba(X)[:, 1]
df["pred_modelo"] = model.predict(X)
df
```

```python
print("Accuracy modelo =", accuracy_score(y, df["pred_modelo"]))
print("Matriz de confusión:")
print(confusion_matrix(y, df["pred_modelo"]))
print()
print(classification_report(y, df["pred_modelo"], zero_division=0))
```
